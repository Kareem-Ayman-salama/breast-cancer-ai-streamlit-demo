````python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image, Markdown

ROOT = Path(r"E:\Freelance work\nour")
OUT = ROOT / "outputs"

def load_json(path):
    path = Path(path)
    if not path.exists():
        print(f"Missing: {path}")
        return {}
    return json.loads(path.read_text(encoding="utf-8"))

def read_csv(path, nrows=None):
    path = Path(path)
    if not path.exists():
        print(f"Missing: {path}")
        return pd.DataFrame()
    return pd.read_csv(path, nrows=nrows)

def show_image(path, width=900):
    path = Path(path)
    if path.exists():
        display(Image(filename=str(path), width=width))
    else:
        print(f"Missing image: {path}")

def show_json_metrics(path):
    data = load_json(path)
    flat = {}
    for k, v in data.items():
        if isinstance(v, dict):
            for kk, vv in v.items():
                if isinstance(vv, (int, float, str)):
                    flat[f"{k}.{kk}"] = vv
        elif isinstance(v, (int, float, str)):
            flat[k] = v
    display(pd.DataFrame([flat]).T.rename(columns={0: "value"}))
````

- The strongest internal result is TCGA Cancer vs Normal signature detection.

- External validation on GEO is mixed, which is expected across different cohorts/platforms and should be reported transparently.
